```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Import and filter Serratus results

```python
# import the serratus results table
serratus = pd.read_csv("serratus_output_combined_S3.csv")
# keep pairs of run_id and palm_id with max pident value
idx_max = serratus.groupby(["run_id", "palm_id"])["pident"].idxmax()
df_max_pident = serratus.loc[idx_max].copy()
# keep only those from selected pident range
d50_95 = df_max_pident[(df_max_pident.pident > 50) & (df_max_pident.pident < 95)]

print(f"After filtering (50 < max pident < 95): {d50_95.run_id.nunique()} SRRs remain.")
```

### Visualize selection on a pident distribution plot

```python
# quick plot
fig, ax = plt.subplots(figsize=(12, 6))

ax = serratus['pident'].plot(kind='kde', xlim=[0,110], title=f'SRR distribution({serratus.run_id.nunique()}) based on pident')
ax.axvspan(50, 95, color='lightpink', alpha=0.25)
ax.axvline(50, color='red', linestyle='--')
ax.axvline(95, color='red', linestyle='--')
ax.text(65, 0.04, f'{d50_95.run_id.nunique()} SRR', fontweight = 'bold')

d50= df_max_pident[df_max_pident.pident <= 50]
d95= df_max_pident[df_max_pident.pident >= 95]

ax.text(20, 0.04, f'{d50.run_id.nunique()} SRR')
ax.text(97, 0.04, f'{d95.run_id.nunique()} SRR')
ax.set_xlabel('pident')

fig.savefig('pident_distribution.png', dpi=300, bbox_inches='tight')
```

### Save selected

```python
samples = sorted(d50_95.run_id.unique())
pd.Series(samples, name='samples').to_csv('samples.tsv', sep='\t', index=False)
```
